```python
business = "Kit's Online Shop"

name = input("What is your name? ")

spending = float(input("How much did you spend? "))

student = input("Are you a student? ")

subtotal = spending

if subtotal >= 100:
    discount = subtotal * 15/100

elif subtotal >= 50:
    discount = subtotal * 10/100

elif student == "yes":
    discount = subtotal * 10/100

else:
    discount = 0


if subtotal >= 50:
    delivery = 0

else:
    delivery = 4.99

total = spending + delivery - discount

print(f"--------Receipt of Purchase-------")
print(f"Name = {name}")
print(f"Spent: {spending}")
print(f"Discount = £{discount:.2f}")
print(f"Delivery = £{delivery:.2f}")
print(f"Total £{total:.2f}")
print(f"Thanks for trusting {business}, see you again soon!")
print("-----------------------------------")
```
