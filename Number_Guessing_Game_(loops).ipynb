```python
secret_number = 7

attempts = 0

guess = int(input("You have three guesses - Guess a number! "))

while guess != secret_number and attempts < 2:

   if guess > secret_number:
     guess = int(input("Too high - Guess again "))
     attempts = attempts + 1

   elif guess < secret_number:
     guess = int(input("Too Low - Guess again "))
     attempts = attempts + 1

if guess == secret_number:
  print("Guessed it!")

else:
  print("Too many guesses - game over! ")
```
